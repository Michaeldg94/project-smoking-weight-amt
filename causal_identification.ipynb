## 1. Formal Causal Question and Identification

Before starting the estimation, we want to precisely define the causal question, introduce potential outcomes notation, and state the assumptions under which the Average Treatment Effect (ATE) is identified from the observed NHEFS data.

### 1.1 Research question

Our research question is **what is the average causal effect of smoking cessation on weight gain over a ten-year follow-up period?**

More precisely, we seek to estimate the weight change (in kg) from 1971 to 1982 that would be *caused* by quitting smoking, averaged over the NHEFS study population, as opposed to the weight change we would observe if no one had quit.

### 1.2 Variables

We define the following variables, all measured at or before the 1971 baseline interview.

| Symbol | NHEFS variable | Type | Description |
|--------|---------------|------|-------------|
| $D_i$ | `qsmk` | Binary | Treatment: 1 if individual $i$ quit smoking by the 1982 follow-up, 0 otherwise |
| $Y_i$ | `wt82_71` | Continuous | Outcome: weight change in kg from 1971 to 1982 |
| $W_i$ | mix of variables | Mixed | Baseline pre-treatment covariates measured in 1971 containing baseline confounders |

### 1.3 Potential outcomes and the estimand

We adopt the **potential outcomes** (Neyman) framework. For each individual $i$, we define:

- $Y_i(1)$: the weight change individual $i$ *would experience* if they quit smoking ($D_i = 1$)
- $Y_i(0)$: the weight change individual $i$ *would experience* if they continued smoking ($D_i = 0$)

These two quantities are in fact counterfactuals for any given individual, because only one is observed in the real world. As we learned, this is the **fundamental problem of causal inference**.

Our target estimand is the **Average Treatment Effect (ATE)**:

$$\text{ATE} = E\bigl[Y_i(1) - Y_i(0)\bigr]$$

This quantity represents the average difference in weight gain that would be caused by quitting smoking versus continuing to smoke, in the NHEFS study population. A positive ATE indicates that quitting causes weight gain on average, (which as we researched is consistent with the pharmacological literature on nicotine suppression of appetite).

We target the ATE rather than the Average Treatment Effect on the Treated (ATT) because we are interested in the population-level question: what would happen if everyone quit, not only among those who actually did quit.

### 1.4 Important assumptions

The ATE is not directly observable because we never observe both $Y_i(1)$ and $Y_i(0)$ for the same individual. For this reason, we need to make three important assumptions that, taken together, allow us to express the ATE as a function of the observed data distribution $P(Y, D, W)$.

---

#### Assumption 1: SUTVA (Stable Unit Treatment Value Assumption)

**Statement.** For all individuals $i$ and treatment values $d \in \{0,1\}$:

$$Y_i = Y_i(D_i) \quad \text{and} \quad Y_i(d) \text{ is well-defined and unique}$$

This has two components:

1. **No interference**: individual $i$'s potential outcomes do not depend on the treatment received by any other individual $j \neq i$.
2. **Consistency / no multiple versions of treatment**: there is a single, well-defined version of "quitting smoking", so $D_i = 1$ maps to a unique potential outcome $Y_i(1)$.

**Justification.** The outcome of interest, weight change, is a private biological outcome primarily determined by individual physiology, diet, and behavior. While direct interference between units is implausible, we acknowledge that mild indirect interference can't be entirely ruled out, for instance, if a household member quits smoking, shared dietary habits and home food environments may shift in ways that affect another household member's weight. However, given that NHEFS does not sample multiple individuals from the same household and that such spillover effects are likely small in magnitude relative to the direct biological effect of cessation, we treat the no-interference component of SUTVA as a reasonable working assumption. The consistency component is more subtle: `qsmk` is defined as complete abstinence from cigarettes by the 1982 follow-up, which is a reasonably homogeneous treatment. We acknowledge that the *route* to quitting (gradual reduction, nicotine replacement or other methods) may vary and could introduce slight heterogeneity in the treatment, but we treat this as negligible given the measurement of NHEFS.

---

#### Assumption 2: Conditional ignorability

$$Y_i(0),\, Y_i(1) \;\perp\!\!\!\perp\; D_i \mid W_i$$

Conditional on the observed baseline covariates $W_i$, treatment assignment is independent of potential outcomes. In other words, within strata of $W$, individuals who quit smoking and those who did not are exchangeable.

**Justification.** Our justification relies on the causal DAG presented in Section 2.5 below. The nine confounders in $W$ block all non-causal paths between $D$ and $Y$ in the DAG: they capture the main biological determinants of both quitting behavior (smoking history, health-consciousness proxied by exercise) and weight change (baseline weight, age, metabolic factors). No variable in $W$ is a collider or a post-treatment variable, ensuring that conditioning on $W$ does not open spurious paths. Of course, we cannot rule out unmeasured confounding entirely, for instance, unobserved psychological factors (stress, motivation) may jointly affect both quitting and eating behavior. We discuss this limitation in Section 5.

---

#### Assumption 3: Positivity (overlap)
For all covariate values $w$ in the support of $W$:

$$0 < P(D_i = 1 \mid W_i = w) < 1$$

Every type of individual, as characterized by their baseline covariates, must have a positive probability of both quitting and not quitting smoking.

**Justification.** Positivity is required for the ATE to be well-defined across all covariate strata. If some subgroup has zero probability of quitting, the counterfactual $Y(1)$ is undefined for that group and the ATE cannot be identified without extrapolation. We estimate the propensity score $\hat{m}(W) = \hat{P}(D=1 \mid W)$ using logistic regression and inspect its distribution separately for quitters and non-quitters positivity. We report this diagnostic below.

### 1.5 Result

Under Assumptions 1–3, we can **nonparametrically** estimate the ATE from the observed data. Specifically, we use the formula :

$$E\bigl[Y_i(1) - Y_i(0)\bigr] = E_W\Bigl[E[Y_i \mid D_i = 1,\, W_i] - E[Y_i \mid D_i = 0,\, W_i]\Bigr]$$

The right-hand side is a function of the observed distribution $P(Y, D, W)$ only. This result justifies all three strategies that we use in this project:

- **IP Weighting**: reweighting the observations by $1/P(D \mid W)$ to create a population in which $D$ and $W$ are independent.
- **Standardization**: we directly compute the outer expectation over $W$ by fitting $E[Y \mid D, W]$ and standardising.
- **Double ML**: we define semiparametric efficiency by partialling out $W$ from both $Y$ and $D$ via flexible ML models, then estimating $\theta$ by OLS of the residuals, either under a partial linear restriction (PLR) or fully nonparametrically via the interactive regression model (IRM).

# Justifications 

**Justification for the PC algorithm**

As we learned, the PC algorithm is a constraint-based causal discovery method that recovers the Markov equivalence class of the underlying DAG from conditional independence tests in the observed data. We use it for two purposes in our analysis: to corroborate the hand-drawn DAG presented in Section XXX, and to detect colliders in the variable set that would be invisible to purely theory-driven covariate selection.

Our key motivation is that conditioning on a collider opens a previously blocked non-causal path between $D$ and $Y$, introducing bias even when all confounders are correctly included. Since we are in a high-dimensional setting, identifying colliders by domain knowledge which we don't have is impossible because we suspect that the variables interact in non-obvious ways. The PC algorithm detects v-structures $X \rightarrow C \leftarrow Z$ directly from the pattern of conditional independencies in the data, that provides a data-based check on our theoretical assumptions.

Importantly we apply PC under the faithfulness assumption, that every conditional independence in the distribution is reflected by a d-separation in the true DAG. We use the skeleton and orientation rules produced by PC to validate edge directions in our DAG and to flag any variable that appears as a collider on a path between $D$ and $Y$. We exclude such variables from $W$ prior to Double ML estimation.

---

**Justification for the Partial Linear Regression (PLR) model**

We write the outcome equation as:

$$Y_i = \theta D_i + g(W_i) + \varepsilon_i, \qquad E[\varepsilon_i \mid D_i, W_i] = 0$$

where $\theta \in \mathbb{R}$ is the constant ATE parameter of interest and $g: \mathcal{W} \rightarrow \mathbb{R}$ is an unknown nuisance function capturing the effect of baseline covariates on weight gain. The propensity score equation is:

$$D_i = m(W_i) + V_i, \qquad E[V_i \mid W_i] = 0$$

The key restriction of the PLR is that the relationship between treatment and the outcome equation  is **linear and additive**, the effect of quitting smoking on weight gain is assumed to be the same constant $\theta$ for every individual, regardless of their covariate profile. The function $g(W)$ is estimated flexibly by a ML model (e.g. Lasso, Random Forest), removing the need for parametric assumptions on confounding. Finally, identification of $\theta$ follows from the Neyman orthogonality condition: after partialling out $W$ from both $Y$ and $D$, the resulting residuals $\tilde{Y} = Y - \hat{l}(W)$ and $\tilde{D} = D - \hat{m}(W)$ satisfy $E[\tilde{Y} \mid \tilde{D}] = \theta \tilde{D}$, so $\theta$ is recovered by OLS.

The PLR requires 2 high-level assumptions beyond the assumptions acknowledged in Section 1.4. First, we use cross-fitting to use to avoid overfitting bias, achieved here via $K=5$ folds. Second, the linearity restriction $Y - g(W) = \theta D + \varepsilon$ must hold; We assess this by plotting $\tilde{Y}$ against $\tilde{D}$ and testing for nonlinearity in the residual regression.

---

**Justification for the Interactive Regression Model (IRM)**

As we learned, the IRM relaxes the linearity restriction of the PLR and is the appropriate specification when treatment effect heterogeneity is plausible. We write:

$$Y_i = g(D_i, W_i) + \varepsilon_i, \qquad E[\varepsilon_i \mid D_i, W_i] = 0$$

where $g(d, w) = E[Y \mid D=d, W=w]$ is estimated separately for $d=0$ and $d=1$ using flexible ML models. We recover the ATE by standardisation:

$$\theta = E\bigl[g(1, W_i) - g(0, W_i)\bigr] \approx \frac{1}{n} \sum_{i=1}^{n} \bigl[\hat{g}(1, W_i) - \hat{g}(0, W_i)\bigr]$$

The key advantage if the IRM is that it imposes no shape restriction on how the treatment effect varies with $W$, for example, it allows quitting smoking to produce greater weight gain in heavy smokers than in light smokers, or in younger individuals than in older ones, without specifying the functional form of this heterogeneity.

We use Double ML via the same Neyman-orthogonal score as in the PLR, augmented with the propensity score $m(W) = P(D=1 \mid W)$. We again use cross-fitting to remove the bias introduced by using the same data to fit nuisance models and compute the final ATE.

The IRM is our preferred primary specification for three reasons. First, the treatment is binary, which is the natural setting for the IRM. Second, effect heterogeneity is substantively plausible: the metabolic consequences of quitting smoking likely vary with smoking history, baseline weight, and age. We also retain PLR as a robustness check; agreement between the two estimates would suggest that the linearity restriction of the PLR is not severely violated in this dataset.
